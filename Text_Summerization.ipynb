```python
# Extractive Summarization (TextRank)
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer

def extractive_summary(text, num_sentences=3):
    parser = PlaintextParser.from_string(text, Tokenizer("english"))
    summarizer = TextRankSummarizer()
    summary = summarizer(parser.document, num_sentences)
    return " ".join([str(sentence) for sentence in summary])
```

```python
# Abstractive Summarization (T5-small & BART)
from transformers import pipeline
t5_summarizer = pipeline("summarization", model="t5-small")
bart_summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

def abstractive_summary(text, model="t5", max_len=60, min_len=10):
    if model == "t5":
        return t5_summarizer(text, max_length=max_len, min_length=min_len, do_sample=False)[0]['summary_text']
    else:
        return bart_summarizer(text, max_length=max_len, min_length=min_len, do_sample=False)[0]['summary_text']
```

```python
# Evaluation with ROUGE
from rouge_score import rouge_scorer
def evaluate_summary(reference, summary):
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
    return scorer.score(reference, summary)
```

```python

# CONFIGURATION
text = """The rapid growth of digital information across the internet, social media platforms,
 news agencies, and scientific publication databases has drastically increased the need for
 automated methods to process and summarize large volumes of textual data. Individuals and
 organizations frequently encounter situations where it is impractical to manually read and
 interpret lengthy documents in a short period of time. As the amount of available text continues
 to multiply, traditional approaches to information retrieval and content understanding have
 become insufficient. This motivates the development of intelligent systems capable of condensing
 complex textual information into concise, meaningful, and contextually accurate summaries.
 Within this domain, text summarization has emerged as a prominent research area in Natural Language
 Processing (NLP), aiming to generate shorter representations of lengthy text while preserving essential
 ideas, semantics, and coherence."""

# Number of sentences for extractive summary
num_sentences = 2

# Abstractive summary length controls
min_len = 20
max_len = 40
```

```python
# RUN SUMMARIZATION
print("ORIGINAL TEXT")
print("="*60)
print(text, "\n")
```

```python
# --- Extractive ---
print("Extractive Summary (TextRank)")
print("-"*20)
extractive = extractive_summary(text, num_sentences)
print(extractive, "\n")
```

```python
# --- Abstractive (T5) ---
print("Abstractive Summary (T5-small)")
print("-"*40)
t5_sum = abstractive_summary(text, model="t5", min_len=min_len, max_len=max_len)
print(t5_sum, "\n")
```

```python
# --- Abstractive (BART) ---
print("Abstractive Summary (BART)")
print("-"*40)
bart_sum = abstractive_summary(text, model="bart", min_len=min_len, max_len=max_len)
print(bart_sum, "\n")
```

```python
# --- Evaluation Example (TextRank) ---
print("Evaluation Metrics (ROUGE)")
print("-"*40)
scores = evaluate_summary(text, extractive)
for metric, score in scores.items():
    print(f"{metric}: Precision={score.precision:.4f}, Recall={score.recall:.4f}, F1={score.fmeasure:.4f}")
```

```python
# --- Evaluation Example (T5) ---
print("Evaluation Metrics (ROUGE)")
print("-"*40)
scores = evaluate_summary(text, t5_sum)
for metric, score in scores.items():
    print(f"{metric}: Precision={score.precision:.4f}, Recall={score.recall:.4f}, F1={score.fmeasure:.4f}")
```

```python
# --- Evaluation Example (BART) ---
print("Evaluation Metrics (ROUGE)")
print("-"*40)
scores = evaluate_summary(text, bart_sum)
for metric, score in scores.items():
    print(f"{metric}: Precision={score.precision:.4f}, Recall={score.recall:.4f}, F1={score.fmeasure:.4f}")
```
